```python
movies = movies.drop_duplicates().reset_index(drop=True)
```

```python
movies = movies[['title','overview','genres','tagline','vote_average','popularity']]

movies.head()
```

```python
movies = movies.dropna(subset=['title'])
```

```python
movies['overview'] = movies['overview'].fillna('')
```

```python
movies['genres'] = movies['genres'].apply(lambda x :" ".join([i['name'] for i in ast.literal_eval(x)]))
```

```python
movies['tagline'] = movies['tagline'].fillna('')
```

```python
movies.isnull().sum()
```

```python
movies['tag'] = movies['overview'] + " " + movies['genres'] + " " + movies['tagline']
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
    #lowercase
    text = str(text).lower()

    #remove punctuations
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.split()

    #remove stopwords
    words = [word for word in words if word not in stop_words]

    #lemmatize
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)

```

```python
movies['tag'] = movies['tag'].apply(preprocess_text)
```

```python
movies = movies.reset_index(drop= True)
```

```python
indices = pd.Series(movies.index, index = movies['title'])
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(movies['tag'])
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n = 10):
    if title not in indices:
        return ['Movie not found']

    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
    similar_idx = sim_score.argsort()[::-1][1:n+1]

    return movies['title'].iloc[similar_idx]

    
```

```python
recommend('toy story')
```

```python
import pickle


pickle.dump(tfidf_matrix, open('tfidf_matrix.pkl', 'wb'))

pickle.dump(indices, open('indices.pkl','wb'))

movies.to_pickle('movies.pkl')

pickle.dump(tfidf, open('tfidf.pkl', 'wb'))
```
